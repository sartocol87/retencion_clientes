```python
# Convertimos los valores de "TotalCharges" a numéricos, forzando los errores a NaN
totalcharges_numeric = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Valores no convertibles a número:", totalcharges_numeric.isna().sum())
# Mostramos las filas con valores no convertibles
df.loc[totalcharges_numeric.isna(), ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]].head(11)    
```
